```python
with open("B.con", "rb") as f:
    data = f.read(8192)

# Признак бинарника: нулевые байты
is_binary = b'\x00' in data
print("Binary" if is_binary else "Likely text")

# Более точно: доля непечатных байтов
non_printable = sum(1 for b in data if b < 9 or 13 < b < 32)
print(f"Non-printable bytes: {non_printable}/{len(data)} ({100*non_printable/len(data):.1f}%)")
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import RegularPolygon
import math
import pandas as pd
import os

os.makedirs('output', exist_ok=True)

def draw_hex_grid(k, hex_size=1, filename="output/hex_grid.png"):
    fig, ax = plt.subplots(figsize=(10, 10))
    
    data = []
    for q in range(-k, k + 1):
        for r in range(-k, k + 1):
            s = -q - r
            if abs(s) <= k:
                x = hex_size * math.sqrt(3) * (q + r / 2)
                y = hex_size * 3 / 2 * r
                
                if abs(x) < 1e-9 and abs(y) < 1e-9:
                    angle = 0.0
                else:
                    angle = math.degrees(math.atan2(y, x))
                    if angle < 0:
                        angle += 360
                        
                data.append({"q": q, "r": r, "x": round(x, 3), "y": round(y, 3), "angle_deg": round(angle, 3)})
                
                hex_patch = RegularPolygon((x, y), numVertices=6, radius=hex_size, 
                                           orientation=math.pi/6, 
                                           facecolor='#e0f2fe', edgecolor='#0369a1', alpha=0.9, linewidth=1.5)
                ax.add_patch(hex_patch)
                
                ax.text(x, y, f"{angle:.3f}°", ha='center', va='center', 
                        fontsize=6, color='#0f172a', fontweight='bold')
                
    ax.set_aspect('equal')
    ax.autoscale_view()
    plt.axis('off')
    plt.title(f"Шестиугольная сетка, радиус k={k}", fontsize=14, fontweight='bold', pad=20)
    plt.tight_layout()
    plt.savefig(filename, bbox_inches='tight', dpi=300)
    plt.close()
    return data

data = draw_hex_grid(3, filename="output/hex_grid.png")
df = pd.DataFrame(data)
df.to_csv("output/hex_grid_data.csv", index=False)
```
