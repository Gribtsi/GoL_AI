```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import math

# ========== ЧАСТЬ 1: Изучение lab_gamma.py ==========
print("=" * 60)
print("ЧАСТЬ 1: Анализ lab_gamma.py")
print("=" * 60)
print("""
Скрипт lab_gamma.py выполняет гамма-коррекцию цветного изображения:
1. Загружает изображение (tiger.png)
2. Применяет формулу: out = 255 * (in/255)^gamma для каждого канала RGB
3. Сохраняет результат в result.png

Формула гамма-коррекции: I_out = 255 * (I_in / 255)^γ
- γ < 1: осветление изображения
- γ > 1: затемнение изображения
- γ = 1: без изменений
""")

# ========== ЧАСТЬ 2: Гамма-коррекция с LUT из 9 ячеек ==========

# Подпрограмма 1: Формирование LUT по заданной гамме
def create_gamma_lut(gamma, num_bins=9):
    """
    Создает таблицу преобразования (LUT) для гамма-коррекции
    
    Parameters:
    - gamma: коэффициент гамма-коррекции
    - num_bins: количество ячеек в LUT (9 ячеек = 8 интервалов)
    
    Returns:
    - lut: массив значений LUT
    """
    # Опорные точки (0, 32, 64, 96, 128, 160, 192, 224, 255)
    input_values = np.linspace(0, 255, num_bins).astype(np.uint8)
    
    # Применяем гамма-коррекцию к опорным точкам
    lut = np.zeros(num_bins)
    for i, val in enumerate(input_values):
        normalized = val / 255.0
        lut[i] = 255.0 * math.pow(normalized, gamma)
    
    return lut, input_values


# Подпрограмма 2: Гамма-коррекция с помощью LUT
def apply_gamma_lut(image, lut, input_values):
    """
    Применяет гамма-коррекцию к изображению используя LUT с линейной интерполяцией
    
    Parameters:
    - image: входное изображение (numpy array)
    - lut: таблица преобразования
    - input_values: опорные точки LUT
    
    Returns:
    - corrected_image: изображение после гамма-коррекции
    """
    height, width, channels = image.shape
    corrected = np.zeros_like(image, dtype=np.float32)
    
    for c in range(channels):
        channel = image[:, :, c].flatten()
        corrected_channel = np.zeros_like(channel, dtype=np.float32)
        
        for i, pixel_val in enumerate(channel):
            # Находим интервал для интерполяции
            if pixel_val <= input_values[0]:
                corrected_channel[i] = lut[0]
            elif pixel_val >= input_values[-1]:
                corrected_channel[i] = lut[-1]
            else:
                # Линейная интерполяция между двумя ближайшими точками
                idx = np.searchsorted(input_values, pixel_val)
                x0, x1 = input_values[idx-1], input_values[idx]
                y0, y1 = lut[idx-1], lut[idx]
                
                # Линейная интерполяция
                t = (pixel_val - x0) / (x1 - x0)
                corrected_channel[i] = y0 + t * (y1 - y0)
        
        corrected[:, :, c] = corrected_channel.reshape(height, width)
    
    return np.clip(corrected, 0, 255).astype(np.uint8)


# Подпрограмма 3: Определение отличия и расчет PSNR
def calculate_psnr(original, corrected):
    """
    Вычисляет PSNR между двумя изображениями
    
    Parameters:
    - original: оригинальное изображение
    - corrected: скорректированное изображение
    
    Returns:
    - psnr: значение PSNR в дБ
    """
    mse = np.mean((original.astype(np.float32) - corrected.astype(np.float32)) ** 2)
    if mse == 0:
        return float('inf')
    
    max_pixel = 255.0
    psnr = 20 * math.log10(max_pixel / math.sqrt(mse))
    return psnr


def gamma_correction_formula(image, gamma):
    """
    Применяет гамма-коррекцию по формуле (как в lab_gamma.py)
    """
    image_array = np.asarray(image)
    
    height, width, channels = image_array.shape
    corrected = np.zeros_like(image, dtype=np.uint8)
    
    for i in range(height):
        for j in range(width):
            for c in range(channels):
                val = image[i, j, c]
                corrected[i, j, c] = int(255 * math.pow(val / 255.0, gamma))
    
    return corrected


# ========== Тестирование гамма-коррекции ==========
print("\n" + "=" * 60)
print("ЧАСТЬ 2: Гамма-коррекция с LUT (9 ячеек)")
print("=" * 60)

# Создаем тестовое изображение
path = "tiger.png"  # Укажите путь к вашему изображению
test_image_png = Image.open(path)
test_image = np.asarray(test_image_png)

# Диапазон значений гамма
gamma_values = np.linspace(0.5, 2.0, 10)
psnr_values = []

for gamma in gamma_values:
    # Метод 1: По формуле (lab_gamma)
    img_formula = gamma_correction_formula(test_image, gamma)
    
    # Метод 2: С помощью LUT
    lut, input_vals = create_gamma_lut(gamma, num_bins=9)
    img_lut = apply_gamma_lut(test_image, lut, input_vals)
    
    # Вычисляем PSNR
    psnr = calculate_psnr(img_formula, img_lut)
    psnr_values.append(psnr)
    
    print(f"Gamma = {gamma:.2f}, PSNR = {psnr:.2f} дБ")

# Построение графика PSNR(gamma)
plt.figure(figsize=(10, 6))
plt.plot(gamma_values, psnr_values, 'o-', linewidth=2, markersize=8)
plt.xlabel('Gamma (γ)', fontsize=12)
plt.ylabel('PSNR (дБ)', fontsize=12)
plt.title('Зависимость PSNR от значения Gamma\n(погрешность LUT vs формула)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('psnr_vs_gamma.png', dpi=150)
print("\nГрафик сохранен как 'psnr_vs_gamma.png'")
plt.show()

# ========== ЧАСТЬ 3: 3D LUT и трилинейная интерполяция ==========
print("\n" + "=" * 60)
print("ЧАСТЬ 3: Преобразование цвета с помощью 3D LUT")
print("=" * 60)

def load_3d_lut(filename, size=9):
    """
    Загружает 3D LUT из файла
    
    Parameters:
    - filename: имя файла LUT
    - size: размер LUT (9x9x9)
    
    Returns:
    - lut_3d: массив размером (size, size, size, 3)
    """
    with open(filename, 'r') as f:
        data = f.read().strip().split()
    
    # Преобразуем строки в числа
    values = [float(x) for x in data]
    
    # Формируем 3D LUT: (R, G, B, RGB)
    lut_3d = np.array(values).reshape(size, size, size, 3)
    
    return lut_3d


def trilinear_interpolation(lut_3d, r, g, b):
    """
    Трилинейная интерполяция для 3D LUT
    
    Parameters:
    - lut_3d: 3D таблица преобразования (9x9x9x3)
    - r, g, b: значения RGB (0-255)
    
    Returns:
    - (r_out, g_out, b_out): преобразованные значения RGB
    """
    size = lut_3d.shape[0]
    
    # Нормализуем входные значения к индексам LUT
    # Для размера 9: индексы от 0 до 8
    r_idx = (r / 255.0) * (size - 1)
    g_idx = (g / 255.0) * (size - 1)
    b_idx = (b / 255.0) * (size - 1)
    
    # Нижние индексы
    r0 = int(np.floor(r_idx))
    g0 = int(np.floor(g_idx))
    b0 = int(np.floor(b_idx))
    
    # Верхние индексы
    r1 = min(r0 + 1, size - 1)
    g1 = min(g0 + 1, size - 1)
    b1 = min(b0 + 1, size - 1)
    
    # Дробные части для интерполяции
    rd = r_idx - r0
    gd = g_idx - g0
    bd = b_idx - b0
    
    # Получаем 8 угловых точек куба
    c000 = lut_3d[r0, g0, b0]
    c001 = lut_3d[r0, g0, b1]
    c010 = lut_3d[r0, g1, b0]
    c011 = lut_3d[r0, g1, b1]
    c100 = lut_3d[r1, g0, b0]
    c101 = lut_3d[r1, g0, b1]
    c110 = lut_3d[r1, g1, b0]
    c111 = lut_3d[r1, g1, b1]
    
    # Трилинейная интерполяция
    # Интерполяция по R
    c00 = c000 * (1 - rd) + c100 * rd
    c01 = c001 * (1 - rd) + c101 * rd
    c10 = c010 * (1 - rd) + c110 * rd
    c11 = c011 * (1 - rd) + c111 * rd
    
    # Интерполяция по G
    c0 = c00 * (1 - gd) + c10 * gd
    c1 = c01 * (1 - gd) + c11 * gd
    
    # Интерполяция по B
    c = c0 * (1 - bd) + c1 * bd
    
    return c


def apply_3d_lut(image, lut_3d):
    """
    Применяет 3D LUT к изображению
    
    Parameters:
    - image: входное изображение (numpy array, RGB, 0-255)
    - lut_3d: 3D таблица преобразования
    
    Returns:
    - transformed_image: преобразованное изображение
    """
    height, width, channels = image.shape
    output = np.zeros_like(image, dtype=np.float32)
    
    for i in range(height):
        for j in range(width):
            r, g, b = image[i, j]
            
            # Применяем трилинейную интерполяцию
            new_rgb = trilinear_interpolation(lut_3d, r, g, b)
            
            # Денормализуем обратно в диапазон 0-255
            output[i, j] = new_rgb * 255.0
    
    return np.clip(output, 0, 255).astype(np.uint8)


# Загружаем и применяем 3D LUT
print("\nЗагружаем 3D LUT из файлов...")

# Используем один из файлов LUT (например, LUT9A.txt)
lut_3d = load_3d_lut('LUT9A.txt', size=9)
print(f"3D LUT загружена. Размер: {lut_3d.shape}")

# Создаем тестовое изображение с градиентом
test_gradient = np.zeros((100, 300, 3), dtype=np.uint8)
for i in range(100):
    for j in range(300):
        test_gradient[i, j] = [j * 255 // 300, i * 255 // 100, 128]

# Применяем 3D LUT
print("Применяем 3D LUT к тестовому изображению...")
transformed_image = apply_3d_lut(test_gradient, lut_3d)

# После коррекции сохраняем
img_corrected = gamma_correction_formula(test_image, gamma=1.5)
Image.fromarray(img_corrected).save('tiger_corrected.png')
print("Сохранено как 'tiger_corrected.png'")

# Открыть файл
Image.open('tiger_corrected.png').show()

# Визуализация результатов
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(test_gradient)
axes[0].set_title('Оригинальное изображение\n(градиент)', fontsize=12)
axes[0].axis('off')

axes[1].imshow(transformed_image)
axes[1].set_title('После применения 3D LUT\n(LUT9A.txt)', fontsize=12)
axes[1].axis('off')

plt.tight_layout()
plt.savefig('3d_lut_transformation.png', dpi=150)
print("Результат сохранен как '3d_lut_transformation.png'")
plt.show()

# ========== Демонстрация всех LUT файлов ==========
print("\n" + "=" * 60)
print("ДОПОЛНИТЕЛЬНО: Применение всех доступных LUT")
print("=" * 60)

lut_files = ['LUT9A.txt', 'LUT9B.txt', 'LUT9C.txt', 'LUT9D.txt', 
             'LUT9E.txt', 'LUT9F.txt', 'LUT9G.txt', 'LUT9H.txt',
             'LUT9I.txt', 'LUT9J.txt', 'LUT9K.txt', 'LUT9L.txt']

# Создаем простое тестовое изображение
simple_test = np.zeros((50, 50, 3), dtype=np.uint8)
simple_test[:, :, 0] = 200  # Красный
simple_test[:, :, 1] = 100  # Зеленый
simple_test[:, :, 2] = 50   # Синий

fig, axes = plt.subplots(3, 4, figsize=(15, 12))
axes = axes.flatten()

for idx, lut_file in enumerate(lut_files):
    try:
        lut = load_3d_lut(lut_file, size=9)
        result = apply_3d_lut(simple_test, lut)
        
        axes[idx].imshow(result)
        axes[idx].set_title(lut_file.replace('.txt', ''), fontsize=10)
        axes[idx].axis('off')
        print(f"✓ {lut_file} применена успешно")
    except Exception as e:
        print(f"✗ Ошибка при обработке {lut_file}: {e}")
        axes[idx].text(0.5, 0.5, f'Ошибка\n{lut_file}', 
                      ha='center', va='center', transform=axes[idx].transAxes)
        axes[idx].axis('off')

plt.tight_layout()
plt.savefig('all_luts_comparison.png', dpi=150)
print("\nСравнение всех LUT сохранено как 'all_luts_comparison.png'")
plt.show()

print("\n" + "=" * 60)
print("ВЫПОЛНЕНИЕ ЗАВЕРШЕНО")
print("=" * 60)

```
