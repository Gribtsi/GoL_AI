```python
import h5py
import numpy as np
import pandas as pd
import sys
import os
from statsmodels.stats.proportion import proportion_confint

HDF5_PATH = r".\shared_data\dataset.h5"
OUT_DIR   = r".\test1\final_shit"

def bayesian_winrate(score_sum, n, k=2):
    """
    Сглаживание к 50% через симметричный Beta prior.
    Эквивалентно prior Beta(k, k), то есть:
      posterior mean = (score_sum + k) / (n + 2*k)

    score_sum:
      сумма black_score, где
      black win = 1.0
      draw      = 0.5
      black loss= 0.0

    n:
      число игр в ячейке (komi, agent)

    k:
      сила prior на каждую сторону.
      k=5 => как будто заранее было 5 "успехов" и 5 "неуспехов",
      то есть 10 виртуальных игр с центром в 50%.
    """
    return (score_sum + k) / (n + 2 * k) * 100.0

os.makedirs(OUT_DIR, exist_ok=True)

# --- 1. Загрузка из HDF5 ---
with h5py.File(HDF5_PATH, "r") as f:
    logs = f["game_logs"]

    def load(name, decode=False):
        data = logs[name][:]
        if decode:
            data = np.array([x.decode("utf-8") for x in data])
        return data

    agent_meta = load("agent_meta", decode=True)
    winner = load("winner")
    black_komi = load("black_komi")
    white_komi = load("white_komi")

# --- 2. Базовые поля ---
komi = white_komi - black_komi

# Победа черных = 1.0, ничья = 0.5, победа белых = 0.0
black_score = np.where(winner == 1, 1.0,
              np.where(winner == 0, 0.5, 0.0))

df = pd.DataFrame({
    "agent": agent_meta,
    "komi": komi,
    "black_score": black_score,
})

# Сохраняем порядок агентов по первому появлению
_, first_idx = np.unique(agent_meta, return_index=True)
agent_order = agent_meta[np.sort(first_idx)]

# --- 3. Агрегация по (komi, agent) ---
grouped = (
    df.groupby(["komi", "agent"], sort=True)["black_score"]
      .agg(score_sum="sum", games="size", raw_winrate="mean")
      .reset_index()
)

grouped["raw_winrate"] = grouped["raw_winrate"] * 100.0
grouped["bayes_winrate"] = bayesian_winrate(
    grouped["score_sum"],
    grouped["games"],
    k=5
)

# --- 4. Pivot raw ---
pivot_raw = grouped.pivot(index="komi", columns="agent", values="raw_winrate")
pivot_raw.index.name = "komi"
pivot_raw.columns.name = None
pivot_raw = pivot_raw[[a for a in agent_order if a in pivot_raw.columns]].sort_index()
pivot_raw = pivot_raw.round(2)

raw_csv = os.path.join(OUT_DIR, "winrate_black_raw.csv")
pivot_raw.to_csv(raw_csv)

# --- 5. Pivot bayes ---
pivot_bayes = grouped.pivot(index="komi", columns="agent", values="bayes_winrate")
pivot_bayes.index.name = "komi"
pivot_bayes.columns.name = None
pivot_bayes = pivot_bayes[[a for a in agent_order if a in pivot_bayes.columns]].sort_index()
pivot_bayes = pivot_bayes.round(2)

bayes_csv = os.path.join(OUT_DIR, "winrate_black_bayes.csv")
pivot_bayes.to_csv(bayes_csv)

print(f"Saved: {raw_csv}")
print(pivot_raw.to_string())

print()
print(f"Saved: {bayes_csv}")
print(pivot_bayes.to_string())
```

```python
import os
from collections import defaultdict, OrderedDict

import h5py
import numpy as np
import pandas as pd
from config import BOARD_SIZE


HDF5_PATH = r".\shared_data\dataset.h5"
OUT_DIR   = r".\test1\final_shit"
CENTER_N = 3
OPENING_PLIES = 8
LONG_GAME_THRESHOLD = 150


GAMES_BLOCK = 200_000   # можно уменьшить/увеличить под RAM
PRINT_TABLES = True


def get_center_cells(board_size: int, center_n: int = 3):
    start = (board_size - center_n) // 2
    end = start + center_n
    cells = set()
    for x in range(start, end):
        for y in range(start, end):
            cells.add(x * board_size + y)
    return cells


def is_board_move_in_center(encoded_move: int, board_size_sqr: int, center_cells):
    m = int(encoded_move)
    if 0 <= m < board_size_sqr:
        return m in center_cells
    if board_size_sqr <= m < 2 * board_size_sqr:
        return (m - board_size_sqr) in center_cells
    return False


def ensure_dir(path: str):
    os.makedirs(path, exist_ok=True)


def decode_agents(raw_arr):
    # raw_arr: numpy array bytes/string from HDF5 slice
    if raw_arr.dtype.kind == "S":
        return [x.decode("utf-8") for x in raw_arr]
    if raw_arr.dtype.kind == "O":
        out = []
        for x in raw_arr:
            if isinstance(x, bytes):
                out.append(x.decode("utf-8"))
            else:
                out.append(str(x))
        return out
    return [str(x) for x in raw_arr]


def update_game_aggs(agg_games, agent_order_seen, agents, komi, length, margin):
    for a, k, l, m in zip(agents, komi, length, margin):
        key = (float(k), a)

        if a not in agent_order_seen:
            agent_order_seen[a] = None

        slot = agg_games[key]
        slot["count"] += 1
        slot["sum_length"] += float(l)
        slot["sum_margin"] += float(m)

def update_long_game_aggs(agg_long_games, agent_order_seen, agents, komi, length, threshold):
    for a, k, l in zip(agents, komi, length):
        key = (float(k), a)
    
        if a not in agent_order_seen:
            agent_order_seen[a] = None
    
        slot = agg_long_games[key]
        slot["count"] += 1
        if float(l) > threshold:
            slot["long_count"] += 1

def build_game_offsets(length_ds, total_games, block_size):
    offsets = np.empty(total_games + 1, dtype=np.int64)
    offsets[0] = 0
    pos = 0
    running = 0

    for start in range(0, total_games, block_size):
        end = min(start + block_size, total_games)
        lengths = np.empty(end - start, dtype=length_ds.dtype)
        length_ds.read_direct(lengths, np.s_[start:end], np.s_[:])
        csum = np.cumsum(lengths, dtype=np.int64)
        offsets[pos + 1: pos + 1 + len(csum)] = running + csum
        running += int(csum[-1]) if len(csum) else 0
        pos += len(csum)

    return offsets


def finalize_pivot(agg_games, agent_order):
    rows = []
    for (k, a), v in agg_games.items():
        cnt = v["count"]
        rows.append(
            {
                "komi": k,
                "agent": a,
                "avg_length": v["sum_length"] / cnt,
                "avg_margin": v["sum_margin"] / cnt,
                "count_games": cnt,
            }
        )

    stats_df = pd.DataFrame(rows)

    if stats_df.empty:
        empty = pd.DataFrame(index=pd.Index([], name="komi"))
        return empty, empty, empty

    ordered_agents = [a for a in agent_order if a in set(stats_df["agent"])]

    pivot_length = (
        stats_df.pivot(index="komi", columns="agent", values="avg_length")
        .reindex(columns=ordered_agents)
        .sort_index()
    )
    pivot_length.index.name = "komi"
    pivot_length.columns.name = None

    pivot_margin = (
        stats_df.pivot(index="komi", columns="agent", values="avg_margin")
        .reindex(columns=ordered_agents)
        .sort_index()
    )
    pivot_margin.index.name = "komi"
    pivot_margin.columns.name = None

    count_pivot = (
        stats_df.pivot(index="komi", columns="agent", values="count_games")
        .reindex(columns=ordered_agents)
        .sort_index()
    )
    count_pivot.index.name = "komi"
    count_pivot.columns.name = None

    pct_games = count_pivot.div(count_pivot.sum(axis=0), axis=1) * 100
    pct_games.index.name = "komi"
    pct_games.columns.name = None

    return pivot_length, pivot_margin, pct_games


def finalize_turns_pivot(agg_turns, agent_order):
    rows = []
    for (k, a), v in agg_turns.items():
        total = v["total_moves"]
        pct = (v["range_moves"] / total * 100.0) if total else np.nan
        rows.append(
            {
                "komi": k,
                "agent": a,
                "pct_81_161": pct,
            }
        )

    stats_df = pd.DataFrame(rows)

    if stats_df.empty:
        return pd.DataFrame(index=pd.Index([], name="komi"))

    ordered_agents = [a for a in agent_order if a in set(stats_df["agent"])]

    pivot = (
        stats_df.pivot(index="komi", columns="agent", values="pct_81_161")
        .reindex(columns=ordered_agents)
        .sort_index()
    )
    pivot.index.name = "komi"
    pivot.columns.name = None
    return pivot

def finalize_long_games_pivot(agg_long_games, agent_order):
    rows = []
    for (k, a), v in agg_long_games.items():
        total = v["count"]
        pct = (v["long_count"] / total * 100.0) if total else np.nan
        rows.append(
            {
                "komi": k,
                "agent": a,
                "pct_games_gt_100": pct,
            }
        )

    stats_df = pd.DataFrame(rows)

    if stats_df.empty:
        return pd.DataFrame(index=pd.Index([], name="komi"))

    ordered_agents = [a for a in agent_order if a in set(stats_df["agent"])]

    pivot = (
        stats_df.pivot(index="komi", columns="agent", values="pct_games_gt_100")
        .reindex(columns=ordered_agents)
        .sort_index()
    )
    pivot.index.name = "komi"
    pivot.columns.name = None
    return pivot


def main():
    ensure_dir(OUT_DIR)

    agg_games = defaultdict(lambda: {
        "count": 0,
        "sum_length": 0.0,
        "sum_margin": 0.0,
    })

    agg_turns = defaultdict(lambda: {
        "total_moves": 0,
        "range_moves": 0,
    })

    agg_center_opening = defaultdict(lambda: {
    "total_opening_board_moves": 0,
    "center_opening_moves": 0,
    })

    agg_long_games = defaultdict(lambda: {
        "count": 0,
        "long_count": 0,
    })


    agent_order_seen = OrderedDict()

    with h5py.File(HDF5_PATH, "r") as f:
        logs = f["game_logs"]

        agent_ds = logs["agent_meta"]
        black_ds = logs["black_komi"]
        white_ds = logs["white_komi"]
        length_ds = logs["length"]
        margin_ds = logs["margin"]
        move_ds = logs["move_flat"]

        total_games = length_ds.shape[0]
        board_size_sqr = BOARD_SIZE * BOARD_SIZE
        center_cells = get_center_cells(BOARD_SIZE, CENTER_N)

        
        # ---- Pass 1: агрегаты по партиям + порядок агентов ----
        for start in range(0, total_games, GAMES_BLOCK):
            end = min(start + GAMES_BLOCK, total_games)
            n = end - start

            black = np.empty(n, dtype=black_ds.dtype)
            white = np.empty(n, dtype=white_ds.dtype)
            length = np.empty(n, dtype=length_ds.dtype)
            margin = np.empty(n, dtype=margin_ds.dtype)

            black_ds.read_direct(black, np.s_[start:end], np.s_[:])
            white_ds.read_direct(white, np.s_[start:end], np.s_[:])
            length_ds.read_direct(length, np.s_[start:end], np.s_[:])
            margin_ds.read_direct(margin, np.s_[start:end], np.s_[:])

            raw_agents = agent_ds[start:end]
            agents = decode_agents(raw_agents)

            komi = white - black

            update_game_aggs(
                agg_games=agg_games,
                agent_order_seen=agent_order_seen,
                agents=agents,
                komi=komi,
                length=length,
                margin=margin,
            )
            
            update_long_game_aggs(
                agg_long_games=agg_long_games,
                agent_order_seen=agent_order_seen,
                agents=agents,
                komi=komi,
                length=length,
                threshold=LONG_GAME_THRESHOLD,
            )

        agent_order = list(agent_order_seen.keys())

        # offsets нужен, чтобы быстро находить диапазон ходов каждой партии
        game_offsets = build_game_offsets(length_ds, total_games, GAMES_BLOCK)

        # ---- Pass 2: агрегаты по ходам, блоками по партиям ----
        for start in range(0, total_games, GAMES_BLOCK):
            end = min(start + GAMES_BLOCK, total_games)
            n = end - start

            black = np.empty(n, dtype=black_ds.dtype)
            white = np.empty(n, dtype=white_ds.dtype)
            length = np.empty(n, dtype=length_ds.dtype)

            black_ds.read_direct(black, np.s_[start:end], np.s_[:])
            white_ds.read_direct(white, np.s_[start:end], np.s_[:])
            length_ds.read_direct(length, np.s_[start:end], np.s_[:])

            raw_agents = agent_ds[start:end]
            agents = decode_agents(raw_agents)
            komi = white - black

            move_start = int(game_offsets[start])
            move_end = int(game_offsets[end])

            moves_block = np.empty(move_end - move_start, dtype=move_ds.dtype)
            move_ds.read_direct(moves_block, np.s_[move_start:move_end], np.s_[:])

            local_offsets = np.empty(n + 1, dtype=np.int64)
            local_offsets[0] = 0
            local_offsets[1:] = np.cumsum(length, dtype=np.int64)

            for i in range(n):
                a = agents[i]
                k = float(komi[i])

                s = int(local_offsets[i])
                e = int(local_offsets[i + 1])

                chunk = moves_block[s:e]
                total_moves = e - s
                range_moves = int(np.count_nonzero((chunk >= 81) & (chunk <= 161)))

                opening_chunk = chunk[:OPENING_PLIES]
                opening_board_mask = (opening_chunk >= 0) & (opening_chunk < 2 * board_size_sqr)
                opening_board_moves = opening_chunk[opening_board_mask]

                center_opening_moves = sum(
                    is_board_move_in_center(m, board_size_sqr, center_cells)
                    for m in opening_board_moves
                )

                slot_center = agg_center_opening[(k, a)]
                slot_center["total_opening_board_moves"] += int(len(opening_board_moves))
                slot_center["center_opening_moves"] += int(center_opening_moves)

                
                slot = agg_turns[(k, a)]
                slot["total_moves"] += total_moves
                slot["range_moves"] += range_moves

    pivot_length, pivot_margin, pct_games = finalize_pivot(agg_games, agent_order)
    pct_moves = finalize_turns_pivot(agg_turns, agent_order)

    pct_center_opening = finalize_turns_pivot(
    defaultdict(lambda: {"total_moves": 0, "range_moves": 0}, {
        key: {
            "total_moves": val["total_opening_board_moves"],
            "range_moves": val["center_opening_moves"],
        }
        for key, val in agg_center_opening.items()
    }), agent_order,)
    
    pct_long_games = finalize_long_games_pivot(agg_long_games, agent_order)


    path1 = os.path.join(OUT_DIR, "avg_game_length.csv")
    path2 = os.path.join(OUT_DIR, "avg_margin.csv")
    path3 = os.path.join(OUT_DIR, "pct_moves_81_161.csv")
    path4 = os.path.join(OUT_DIR, "pct_games_by_komi.csv")
    path5 = os.path.join(OUT_DIR, "pct_opening_board_moves_center_3x3.csv")
    path6 = os.path.join(OUT_DIR, "pct_games_gt_150.csv")

    
    pivot_length.to_csv(path1)
    pivot_margin.to_csv(path2)
    pct_moves.to_csv(path3)
    pct_games.to_csv(path4)
    pct_center_opening.to_csv(path5)
    pct_long_games.to_csv(path6)


    if PRINT_TABLES:
        pd.set_option("display.max_rows", None)
        pd.set_option("display.max_columns", None)
        pd.set_option("display.width", 200)

        print("=== Avg Game Length ===")
        print(pivot_length.to_string())

        print("\n=== Avg Margin (score diff) ===")
        print(pivot_margin.to_string())

        print("\n=== % Moves with move_flat in [81..161] ===")
        print(pct_moves.to_string())

        print("\n=== % Games per (komi, agent) column-normalized ===")
        print(pct_games.to_string())

        print("\n=== % Opening board moves in center 3x3 (first 6 plies only) ===")
        print(pct_center_opening.to_string())

        print("\n=== % Games with length > 100 ===")
        print(pct_long_games.to_string())

main()
```

```python
import os
import csv
from collections import defaultdict
from  board import BOARD_SIZE

import h5py
import numpy as np


HDF5_PATH = r".\shared_data\dataset.h5"     # твой путь
OUT_DIR   = r".\test1\final_shit"
OUT_CSV   = "opening_heatmaps_first10_fullmoves.csv"

GAMES_BLOCK = 200_000          # как и раньше
LAST_MOVES = 40_000_000        # последние 40M ходов
NUM_FULL_TURNS = 20

def ensure_dir(path: str):
    os.makedirs(path, exist_ok=True)


def build_game_offsets(length_ds, total_games, block_size):
    offsets = np.empty(total_games + 1, dtype=np.int64)
    offsets[0] = 0
    pos = 0
    running = 0

    for start in range(0, total_games, block_size):
        end = min(start + block_size, total_games)
        n = end - start
        buf = np.empty(n, dtype=length_ds.dtype)
        length_ds.read_direct(buf, np.s_[start:end], np.s_[:])
        csum = np.cumsum(buf, dtype=np.int64)
        offsets[pos + 1: pos + 1 + n] = running + csum
        running += int(csum[-1]) if n > 0 else 0
        pos += n

    return offsets


def main():
    ensure_dir(OUT_DIR)

    with h5py.File(HDF5_PATH, "r") as f:
        logs = f["game_logs"]
        move_ds   = logs["move_flat"]   # или "move_flat", если так называется
        length_ds = logs["length"]

        total_games = length_ds.shape[0]

        # Кумулятивные offsets по играм в общем массиве ходов
        game_offsets = build_game_offsets(length_ds, total_games, GAMES_BLOCK)

        total_moves = int(game_offsets[-1])
        if total_moves == 0:
            raise RuntimeError("No moves in dataset")

        # Рабочий диапазон: последние LAST_MOVES ходов
        global_start = max(0, total_moves - LAST_MOVES)
        global_end   = total_moves   # exclusive

        # Счетчики: player -> layer -> move_index(1..10) -> 9x9
        # player: 0=black, 1=white
        counts = {
            0: {0: np.zeros((NUM_FULL_TURNS, BOARD_SIZE, BOARD_SIZE), dtype=np.int64), 1: np.zeros((NUM_FULL_TURNS, BOARD_SIZE, BOARD_SIZE), dtype=np.int64)},
            1: {0: np.zeros((NUM_FULL_TURNS, BOARD_SIZE, BOARD_SIZE), dtype=np.int64), 1: np.zeros((NUM_FULL_TURNS, BOARD_SIZE, BOARD_SIZE), dtype=np.int64)},
        }

        # проходим по партиям блоками, как обычно
        for start in range(0, total_games, GAMES_BLOCK):
            end = min(start + GAMES_BLOCK, total_games)
            n_games = end - start
            if n_games <= 0:
                continue

            # читаем длины партий блока
            length_buf = np.empty(n_games, dtype=length_ds.dtype)
            length_ds.read_direct(length_buf, np.s_[start:end], np.s_[:])

            # смещения ходов в общем массиве для блока
            block_global_offsets = game_offsets[start:end+1]  # len = n_games+1

            # проверяем, пересекается ли этот блок с [global_start, global_end)
            block_first = int(block_global_offsets[0])
            block_last  = int(block_global_offsets[-1])
            if block_last <= global_start or block_first >= global_end:
                continue  # блок целиком вне интересующего диапазона

            # читаем все ходы блока, которые потенциально пересекают диапазон
            read_start = max(global_start, block_first)
            read_end   = min(global_end, block_last)
            if read_end <= read_start:
                continue

            block_moves = np.empty(read_end - read_start, dtype=move_ds.dtype)
            move_ds.read_direct(block_moves, np.s_[read_start:read_end], np.s_[:])

            # для маппинга: global_index -> local_index
            # local = global - read_start
            # для каждой партии в блоке
            for gi in range(n_games):
                g_begin = int(block_global_offsets[gi])
                g_end   = int(block_global_offsets[gi + 1])

                # пересечение партии с глобальным диапазоном
                inter_start = max(g_begin, global_start)
                inter_end   = min(g_end,   global_end)
                if inter_end <= inter_start:
                    continue  # эта партия не пересекает интересующие последние ходы

                # локальные индексы в block_moves
                local_start = inter_start - read_start
                local_end   = inter_end   - read_start
                game_moves  = block_moves[local_start:local_end]

                # нам нужен "дебют" — первые 10 ходов каждого игрока
                # делаем проход, считая ходы черных/белых
                black_seen = 0
                white_seen = 0

                for ply, mv in enumerate(game_moves):
                    # кто ходит: считаем, что индекс 0 — черные, 1 — белые и т.д.
                    player = ply % 2  # 0=black, 1=white

                    # учитываем только ходы на доске (0..161), без пассов/резигнов
                    if mv < 0 or mv >= BOARD_SIZE * BOARD_SIZE * 2:
                        continue

                    # индекс хода данного игрока
                    if player == 0:
                        black_seen += 1
                        move_idx = black_seen
                    else:
                        white_seen += 1
                        move_idx = white_seen

                    if move_idx > NUM_FULL_TURNS:
                        # нас интересуют только первые 10 ходов этого игрока
                        continue

                    # слой и клетка
                    if mv < BOARD_SIZE * BOARD_SIZE:
                        layer = 0  # установка камней
                        pos   = int(mv)
                    else:
                        layer = 1  # цикл жизни
                        pos   = int(mv - BOARD_SIZE * BOARD_SIZE)

                    row = pos // BOARD_SIZE
                    col = pos % BOARD_SIZE

                    # move_idx в массиве от 0 до 9
                    counts[player][layer][move_idx - 1, row, col] += 1

        # ---- Запись в CSV ----
        out_path = os.path.join(OUT_DIR, OUT_CSV)
        with open(out_path, "w", newline="", encoding="utf-8") as f_csv:
            writer = csv.writer(f_csv)
            writer.writerow(["player", "layer", "move_index", "row", "col", "count"])

            for player in (0, 1):
                player_name = "black" if player == 0 else "white"
                for layer in (0, 1):
                    layer_name = "place" if layer == 0 else "life"

                    grid = counts[player][layer]  # shape (10, 9, 9)
                    for mi in range(NUM_FULL_TURNS):          # 0..9 => move_index 1..10
                        move_index = mi + 1
                        sub = grid[mi]
                        # Можно писать все клетки, даже с нулем,
                        # или только ненулевые – я буду писать все, чтобы удобно было читать.
                        for r in range(BOARD_SIZE):
                            for c in range(BOARD_SIZE):
                                cnt = int(sub[r, c])
                                writer.writerow(
                                    [player_name, layer_name, move_index, r, c, cnt]
                                )

    print("Saved opening heatmap stats to:", out_path)

main()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize


def text_color_for_value(val):
    r, g, b, _ = cmap_obj(norm(val))
    luminance = 0.2126 * r + 0.7152 * g + 0.0722 * b
    return "black" if luminance > 0.5 else "white"


def plot_opening_heatmaps_2x2(
    move_index: int,
    csv_path: str = r".\test1\final_shit\opening_heatmaps_first10_fullmoves.csv",
    cmap: str = "YlOrRd",
    annotate: bool = True,
    save_path: str | None = None,
):
    """
    Рисует 4 хитмапы для одного полного хода (1..10):

    | камень черных      | камень белых      |
    | цикл жизни черных  | цикл жизни белых  |

    Единственная легенда справа — частотность в процентах.
    """

    df = pd.read_csv(csv_path)

    required_cols = {"player", "layer", "move_index", "row", "col", "count"}
    missing = required_cols - set(df.columns)
    if missing:
        raise ValueError(f"CSV missing columns: {missing}")

    df = df[df["move_index"] == move_index].copy()
    if df.empty:
        raise ValueError(f"No data found for move_index={move_index}")

    layout = [
        ("black", "place", "Черные\nУстановка камня"),
        ("white", "place", "Белые\nУстановка камня"),
        ("black", "life",  "Цикл жизни"),
        ("white", "life",  "Цикл жизни"),
    ]

    def make_grid(player: str, layer: str) -> np.ndarray:
        sub = df[(df["player"] == player) & (df["layer"] == layer)]
        if sub.empty:
            return np.zeros((BOARD_SIZE, BOARD_SIZE), dtype=np.float64)

        grid = np.zeros((BOARD_SIZE, BOARD_SIZE), dtype=np.float64)
        rows = sub["row"].to_numpy(dtype=np.int64)
        cols = sub["col"].to_numpy(dtype=np.int64)
        vals = sub["count"].to_numpy(dtype=np.float64)
        grid[rows, cols] = vals

        return grid

    black_place = make_grid("black", "place")
    black_life  = make_grid("black", "life")
    white_place = make_grid("white", "place")
    white_life  = make_grid("white", "life")

    black_total = black_place.sum() + black_life.sum()
    white_total = white_place.sum() + white_life.sum()

    
    black_place_pct = (black_place / black_total * 100.0) if black_total > 0 else black_place
    black_life_pct  = (black_life  / black_total * 100.0) if black_total > 0 else black_life
    white_place_pct = (white_place / white_total * 100.0) if white_total > 0 else white_place
    white_life_pct  = (white_life  / white_total * 100.0) if white_total > 0 else white_life

    grids = [
    black_place_pct,
    white_place_pct,
    black_life_pct,
    white_life_pct,]

    vmax = max(float(g.max()) for g in grids)
    vmax = max(vmax, 1e-12)

    fig, axes = plt.subplots(2, 2, figsize=(10, 9), constrained_layout=True)
    norm = Normalize(vmin=0.0, vmax=vmax)
    
    
    cmap_obj = plt.get_cmap(cmap)
    def text_color_for_value(val):
        r, g, b, _ = cmap_obj(norm(val))
        luminance = 0.2126 * r + 0.7152 * g + 0.0722 * b
        return "black" if luminance > 0.5 else "white"

        
    ims = []
    for ax, grid, (_, _, title) in zip(axes.ravel(), grids, layout):
        im = ax.imshow(grid, cmap=cmap, norm=norm, origin="upper")
        ims.append(im)

        ax.set_title(title, fontsize=12)
        ax.set_xticks(range(9))
        ax.set_yticks(range(9))
        ax.set_xticklabels(range(9))
        ax.set_yticklabels(range(9))

        # сетка 9x9
        ax.set_xticks(np.arange(-0.5, 9, 1), minor=True)
        ax.set_yticks(np.arange(-0.5, 9, 1), minor=True)
        ax.grid(which="minor", color="white", linestyle="-", linewidth=0.8, alpha=0.35)
        ax.tick_params(which="minor", bottom=False, left=False)


        if annotate:
            for r in range(BOARD_SIZE):
                for c in range(BOARD_SIZE):
                    val = grid[r, c]
                    if val <= 1:
                        continue
                    color = text_color_for_value(val)
                    ax.text(
                        c, r, f"{val:.1f}%",
                        ha="center", va="center",
                        fontsize=8, color=color
                    )

    cbar = fig.colorbar(ims[0], ax=axes.ravel().tolist(), shrink=0.92, pad=0.03)
    cbar.set_label("Частотность, %", rotation=90)

    #fig.suptitle(f"Дебютные хитмапы, полный ход #{move_index}", fontsize=14)

    if save_path is not None:
        fig.savefig(save_path, dpi=180, bbox_inches="tight")

    plt.show()
    return fig, axes
```

```python
plot_opening_heatmaps_2x2(
    move_index=7,
    csv_path=r".\test1\final_shit\opening_heatmaps_first10_fullmoves.csv"
)
```
