```python
from IPython.display import clear_output
import time

def run_test_scenario(commands, delay=1.0, print_score = False):
    """
    Универсальный метод для запуска тестовых сценариев.
    
    Args:
        commands: Список команд в формате:
            - Move((x, y), life_cycles, color) - выполнить ход
            - ('set', x, y, color) - напрямую установить камень на доске (для подготовки)
            - ('clear',) - очистить доску
            - ('comment', text) - вывести комментарий
        delay: Задержка между командами в секундах
    
    Примеры команд:
        Move((9, 9), 0, Board.BLACK)  # Черные ставят камень в (9,9) без цикла жизни
        Move((10, 10), 1, Board.WHITE)  # Белые ставят камень в (10,10) с циклом жизни
        Move(None, 0, Board.BLACK)  # Черные пасуют
        ('set', 5, 5, Board.BLACK)  # Напрямую установить черный камень
        ('clear',)  # Очистить доску
        ('comment', 'Тест захвата')  # Комментарий
    """
    # Инициализация доски
    board = Board()
    log_outputs = []
    for i, cmd in enumerate(commands):
        clear_output(wait=True)
        
        # Проверяем тип команды
        if isinstance(cmd, Move):
            # Это объект Move - выполняем ход
            print(f"Шаг {i+1}: {cmd}")
            result = board.make_move(cmd)
            success = result[0]
            message = result[1]
            
            if not success:
                log_outputs.append(f"❌ ХОД ОТКЛОНЕН: {cmd}\n{message}\n")
            
            print(board.display_board())
            for i in log_outputs:
                print(i)
        elif isinstance(cmd, tuple):
            cmd_type = cmd[0]
            
            if cmd_type == 'set':
                # Напрямую установить камень (для подготовки сценария)
                x, y, color = cmd[1], cmd[2], cmd[3]
                board.current_state[x, y] = color
                print(f"Шаг {i+1}: Установлен камень напрямую в ({x}, {y}), цвет={color}")
                print(board.display_board())
                for i in log_outputs:
                    print(i)
                
            elif cmd_type == 'clear':
                # Очистить доску
                board.current_state.fill(0)
                board.history_stack.clear()
                print(f"Шаг {i+1}: Доска очищена")
                print(board.display_board())
                for i in log_outputs:
                    print(i)
            elif cmd_type == 'comment':
                # Вывести комментарий
                text = cmd[1]
                print("=" * 60)
                print(f"  {text}")
                print("=" * 60)
                print(board.display_board())
                for i in log_outputs:
                    print(i)
        
        time.sleep(delay)
    
    # Финальное состояние
    clear_output(wait=True)
    print("\n" + "=" * 60)
    print("ТЕСТ ЗАВЕРШЕН")
    print("=" * 60)
    print(board.display_board())                
    for i in log_outputs:
        print(i)

    if print_score:
        board.print_score()
    
    return board
```
