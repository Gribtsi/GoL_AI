```python
with open("file.txt", "rb") as f:
    data = f.read(8192)

# Признак бинарника: нулевые байты
is_binary = b'\x00' in data
print("Binary" if is_binary else "Likely text")

# Более точно: доля непечатных байтов
non_printable = sum(1 for b in data if b < 9 or 13 < b < 32)
print(f"Non-printable bytes: {non_printable}/{len(data)} ({100*non_printable/len(data):.1f}%)")
```
